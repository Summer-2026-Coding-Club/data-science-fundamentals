## 1. NumPy Arrays

A NumPy array is like a list, but faster and built for math. Operations apply to every element at once (no loop needed).

```python
ages = np.array([14, 15, 16, 17, 18])
print(ages)
print(ages + 1)        # add 1 to every element
print(ages * 2)        # double every element
print(ages.mean())     # average
print(ages.size)
print(ages.max(), ages.min())
```

## 2. Building a DataFrame

A pandas `DataFrame` is a table — rows and columns, like a spreadsheet. The easiest way to build one is from a dictionary, where each key becomes a column.

```python
students = pd.DataFrame({
    "name": ["Ada", "Sam", "Lin", "Theo", "Maya"],
    "grade": [9, 10, 9, 11, 12],
    "test_score": [88, 72, 95, 60, 81],
})
students
```

```python
# Your code here
movies = pd.DataFrame({
    "title": ["3 Idiots", "The Lord of The Rings", "Dangal", "Jumangi", "Enemy at the Gates", "1917"],
    "genre": ["Comidey", "Fantasy", "Real Story", "Fantasy", "Action", "Action"],
    "rating": [9, 8, 7, 7, 6, 5]
})

songs = pd.DataFrame({
    "title": ["Dai Dai", "Waka Waka", "Alone", "Rockeybi", "Illuminate", "Goals"],
    "genre": ["Jaz", "Classic", "Bob", "Rock", "Rap", "Keys"],
    "rating": [7, 6, 5, 8, 7, 3]
})

print(movies)
print(songs)
```

## 4. Selecting and Filtering

You can grab a single column with `df["column"]`, and filter rows with a condition inside `[]`.

```python
print(students["name"])

high_scorers = students[students["test_score"] > 80]
high_scorers
```

## 5. Adding Columns and Sorting

You can add a new column by assigning to it, and sort with `.sort_values()`.

```python
students["passed"] = students["test_score"] >= 70
students_sorted = students.sort_values("test_score", ascending=False)
students_sorted
```

## 6. Grouping and Aggregating

`.groupby()` splits the table into groups based on a column, so you can compute stats per group (e.g. average score per grade).

```python
students.groupby("grade")["test_score"].mean()
```

**Exercise:** Group your movies/songs DataFrame by `genre` and find the average rating per genre.

```python
# Your code here
average_rating_per_genre = movies.groupby('genre')['rating'].mean()
average_rating_per_genre.sort_values(ascending=False)
```

## 7. A First Plot

Matplotlib turns data into charts. A bar chart is a good default for comparing values across categories.

```python
plt.bar(students["name"], students["test_score"])
plt.xlabel("Student")
plt.ylabel("Test Score")
plt.title("Test Scores by Student")
plt.show()
```
