# Impact of AI on Students
A comprehensive dataset capturing how Generative AI tool usage shapes student performance, skill development, mental health, and burnout risk across 50,000 students.

## About the Dataset
As Generative AI tools like ChatGPT, Copilot, and Gemini become deeply embedded in academic life, understanding their **true impact on students** has never been more critical. This dataset provides a rich, multi-dimensional snapshot of student interactions with AI — spanning academic outcomes, behavioral patterns, institutional context, and psychological well-being.

Whether you're a researcher, educator, data scientist, or policymaker, this dataset offers fertile ground for exploring the **benefits and risks of AI adoption in higher education**.



```python
import pandas as pd
import matplotlib.pyplot as plt
```

## 1. Importing the Data

```python
ai_student_impact = pd.read_csv("data/ai_student_impact_dataset.csv")
ai_student_impact
```

## 2. Exploring the Data

```python
ai_student_impact.describe()
```

```python
average_hours_per_week = ai_student_impact.groupby('Prompt_Engineering_Skill')['Weekly_GenAI_Hours'].mean().sort_values(ascending=False)
average_hours_per_week
```

```python
plt.hist(ai_student_impact['Weekly_GenAI_Hours'])
```

```python
plt.boxplot(ai_student_impact["Tool_Diversity"])
ai_student_impact.columns
```

```python
ai_student_impact['Prompt_Engineering_Skill'].value_counts().plot(kind='bar')
```

```python
plt.scatter(ai_student_impact['Weekly_GenAI_Hours'], ai_student_impact["Post_Semester_GPA"])
plt.xlabel("Weekly GenAI Hours")
plt.ylabel("Post Semester GPA")
plt.title("The relationship between the students' number of hours per week and their GPA")
plt.show()
```

```python
ai_student_impact.groupby("Major_Category")['Post_Semester_GPA'].mean().sort_values(ascending=False)
ai_student_impact.groupby("Year_of_Study")['Post_Semester_GPA'].mean().sort_values(ascending=False)

```
